````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from collections import defaultdict

plt.style.use('seaborn')
plt.rcParams['figure.figsize'] = [12, 6]

class IntradayTimeAnalyzer:
    def __init__(self, data_dict):
        """
        Initialize with dictionary of DataFrames:
        {
            'EURUSD': DataFrame with ['Open','High','Close'] columns and DateTime index,
            'GBPUSD': ...,
            ...
        }
        """
        self.data_dict = data_dict
        self.pairs = list(data_dict.keys())
        self.results = defaultdict(dict)
        
    def _preprocess_data(self):
        """Calculate common metrics for all pairs"""
        for pair in self.pairs:
            df = self.data_dict[pair].copy()
            df['Returns'] = df['Close'].pct_change()
            df['Range'] = df['High'] - df['Open']  # Intraday range
            df['Hour'] = df.index.hour
            df['DayOfWeek'] = df.index.dayofweek  # Monday=0
            self.data_dict[pair] = df

    def analyze_time_patterns(self):
        """Run all time-based analyses"""
        self._preprocess_data()
        
        for pair in self.pairs:
            df = self.data_dict[pair]
            
            # 1. Hourly returns analysis
            hourly = df.groupby('Hour')['Returns'].agg(['mean','std','count'])
            hourly['t_stat'], hourly['p_value'] = zip(*[
                stats.ttest_1samp(group, 0) 
                for _, group in df.groupby('Hour')['Returns']
            ])
            self.results[pair]['hourly'] = hourly
            
            # 2. Day of week analysis
            dow = df.groupby('DayOfWeek')['Returns'].agg(['mean','std'])
            self.results[pair]['day_of_week'] = dow
            
            # 3. Session overlaps (London/NY)
            london_mask = (df['Hour'] >= 8) & (df['Hour'] < 16)
            ny_mask = (df['Hour'] >= 13) & (df['Hour'] < 21)
            overlap_mask = (df['Hour'] >= 13) & (df['Hour'] < 16)
            
            sessions = {
                'London': df[london_mask]['Returns'],
                'New York': df[ny_mask]['Returns'],
                'Overlap': df[overlap_mask]['Returns']
            }
            self.results[pair]['sessions'] = sessions
            
        return self.results
    
    def plot_hourly_patterns(self, pair):
        """Visualize hourly patterns for a currency pair"""
        if not self.results:
            self.analyze_time_patterns()
            
        hourly = self.results[pair]['hourly']
        
        fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True)
        
        # Hourly mean returns with confidence intervals
        ax1.bar(hourly.index, hourly['mean'], 
               yerr=1.96*hourly['std']/np.sqrt(hourly['count']),
               color='steelblue', alpha=0.7)
        ax1.set_title(f'{pair} Hourly Returns Pattern')
        ax1.set_ylabel('Mean Return (%)')
        ax1.axhline(0, color='black', linestyle='--')
        
        # Mark significant hours (p < 0.05)
        sig_hours = hourly[hourly['p_value'] < 0.05].index
        for hour in sig_hours:
            ax1.axvspan(hour-0.5, hour+0.5, color='green', alpha=0.1)
        
        # Hourly volatility
        ax2.plot(hourly.index, hourly['std'], color='firebrick')
        ax2.set_title('Hourly Volatility')
        ax2.set_xlabel('Hour of Day (GMT)')
        ax2.set_ylabel('Standard Deviation')
        
        plt.tight_layout()
        plt.show()
    
    def plot_session_comparison(self, pair):
        """Compare performance across trading sessions"""
        sessions = self.results[pair]['sessions']
        
        fig, ax = plt.subplots()
        session_data = [sessions[k] for k in sessions]
        
        ax.boxplot(session_data, labels=list(sessions.keys()))
        ax.set_title(f'{pair} Session Performance Comparison')
        ax.set_ylabel('Returns')
        ax.axhline(0, color='black', linestyle='--')
        
        # Add significance markers
        _, p_lon_ny = stats.ttest_ind(sessions['London'], sessions['New York'])
        if p_lon_ny < 0.05:
            ax.text(1.5, ax.get_ylim()[1]*0.9, 
                   f'p = {p_lon_ny:.3f}', ha='center')
        
        plt.show()
    
    def plot_day_of_week_heatmap(self):
        """Heatmap of returns by hour and day for all pairs"""
        all_data = []
        
        for pair in self.pairs:
            df = self.data_dict[pair]
            heatmap_data = df.pivot_table(
                index='Hour',
                columns='DayOfWeek',
                values='Returns',
                aggfunc='mean'
            )
            all_data.append((pair, heatmap_data))
        
        fig, axes = plt.subplots(len(self.pairs), 1, figsize=(12, 3*len(self.pairs)))
        
        for idx, (pair, data) in enumerate(all_data):
            cax = axes[idx].imshow(
                data, 
                aspect='auto',
                cmap='RdYlBu',
                vmin=-0.0005, 
                vmax=0.0005
            )
            axes[idx].set_title(pair)
            axes[idx].set_yticks(range(24))
            axes[idx].set_yticklabels(data.index)
            axes[idx].set_xticks(range(5))
            axes[idx].set_xticklabels(['Mon','Tue','Wed','Thu','Fri'])
            
            if idx == len(self.pairs)-1:
                fig.colorbar(cax, ax=axes[idx], label='Mean Return')
        
        plt.tight_layout()
        plt.show()

# Example Usage
if __name__ == "__main__":
    # Create sample data (replace with your actual data)
#     date_rng = pd.date_range('2023-01-01', '2023-03-31', freq='15min')
#     data_dict = {}
    
#     for pair in ['EURUSD', 'GBPUSD', 'USDJPY']:
#         prices = np.cumprod(1 + np.random.normal(0, 0.0001, len(date_rng))) * 1.10
        
#         # Add time-of-day effects
#         hour_effect = np.sin(np.pi * date_rng.hour / 12) * 0.0003
#         prices = prices * (1 + hour_effect)
        
#         data_dict[pair] = pd.DataFrame({
#             'Open': prices,
#             'High': prices + np.abs(np.random.normal(0, 0.0002, len(prices))),
#             'Close': prices
#         }, index=date_rng)
    
    data_dict = dict_curr 
    # Initialize and analyze
    analyzer = IntradayTimeAnalyzer(data_dict)
    results = analyzer.analyze_time_patterns()
    
    # Visualizations
    for pair in data_dict.keys():
        analyzer.plot_hourly_patterns(pair)
        analyzer.plot_session_comparison(pair)
    
    analyzer.plot_day_of_week_heatmap()
    
    # Print key statistics
    print("\nBest Performing Hours:")
    for pair in data_dict.keys():
        best_hour = results[pair]['hourly']['mean'].idxmax()
        print(f"{pair}: Hour {best_hour} GMT - Mean Return {results[pair]['hourly']['mean'][best_hour]:.4%}")
````

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from statsmodels.tsa.seasonal import STL
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
plt.style.use('seaborn')

class IntradayTimePatterns:
    def __init__(self, ohlc_data, session_hours=None):
        """
        Initialize with OHLC data containing datetime index
        ohlc_data: DataFrame with columns ['open', 'high', 'low', 'close']
        session_hours: dict of market sessions {'london': (8,16), ...}
        """
        self.data = ohlc_data.copy()
        self.data['returns'] = self.data['close'].pct_change()
        self.data['log_returns'] = np.log(self.data['close']/self.data['close'].shift(1))
        self.data['range'] = self.data['high'] - self.data['low']
        
        # Default market hours (GMT)
        self.session_hours = session_hours or {
            'tokyo': (0, 6),
            'london': (8, 16),
            'new_york': (13, 21)
        }
        
        # Extract time features
        self._extract_time_features()
        
    def _extract_time_features(self):
        """Create time-based features for analysis"""
        self.data['hour'] = self.data.index.hour
        self.data['minute'] = self.data.index.minute
        self.data['day_of_week'] = self.data.index.dayofweek  # Monday=0
        self.data['session'] = self._assign_sessions()
        
    def _assign_sessions(self):
        """Label each observation with market session"""
        sessions = []
        for idx in self.data.index:
            hour = idx.hour
            if self.session_hours['tokyo'][0] <= hour < self.session_hours['tokyo'][1]:
                sessions.append('tokyo')
            elif self.session_hours['london'][0] <= hour < self.session_hours['london'][1]:
                sessions.append('london')
            elif self.session_hours['new_york'][0] <= hour < self.session_hours['new_york'][1]:
                sessions.append('new_york')
            else:
                sessions.append('overlap' if hour in [6,7,12,21,22] else 'off_hours')
        return sessions
    
    def analyze_hourly_patterns(self, metric='returns'):
        """
        Analyze hourly statistics for specified metric
        Options: 'returns', 'log_returns', 'range', 'volatility'
        """
        if metric == 'volatility':
            values = self.data['returns'].rolling(6).std()  # 30-min volatility
        else:
            values = self.data[metric]
            
        hourly_stats = values.groupby(self.data['hour']).agg(
            ['mean', 'median', 'std', 'count', 
             lambda x: stats.skew(x), 
             lambda x: stats.kurtosis(x)]
        )
        hourly_stats.columns = ['mean', 'median', 'std', 'count', 'skewness', 'kurtosis']
        hourly_stats['sharpe'] = hourly_stats['mean'] / hourly_stats['std']
        
        # Significance testing
        p_values = []
        for hour in hourly_stats.index:
            _, p = stats.ttest_1samp(
                values[self.data['hour'] == hour].dropna(),
                values.mean()
            )
            p_values.append(p)
        hourly_stats['p_value'] = p_values
        
        self.hourly_stats = hourly_stats
        return hourly_stats
    
    def analyze_session_patterns(self):
        """Compare performance across market sessions"""
        session_stats = self.data.groupby('session')['returns'].agg(
            ['mean', 'std', 'count', 'median']
        )
        session_stats['sharpe'] = session_stats['mean'] / session_stats['std']
        
        # ANOVA test for session differences
        groups = [group['returns'].values for name, group in self.data.groupby('session')]
        _, anova_p = stats.f_oneway(*groups)
        
        self.session_stats = session_stats
        self.session_anova_p = anova_p
        return session_stats, anova_p
    
    def decompose_intraday_seasonality(self, period=24*3):
        """STL decomposition of intraday seasonality"""
        stl = STL(
            self.data['close'].dropna(),
            period=period,  # 3-day period for intraday patterns
            seasonal=25,
            robust=True
        )
        res = stl.fit()
        
        self.seasonal = res.seasonal
        self.trend = res.trend
        self.resid = res.resid
        
        return res
    
    def plot_hourly_metrics(self, metric='returns'):
        """Interactive plot of hourly patterns"""
        if not hasattr(self, 'hourly_stats'):
            self.analyze_hourly_patterns(metric)
            
        fig = make_subplots(rows=2, cols=1, shared_xaxes=True,
                           subplot_titles=(f'Hourly {metric.capitalize()} Patterns', 
                                         'Statistical Significance'))
        
        # Mean returns with confidence intervals
        fig.add_trace(
            go.Scatter(
                x=self.hourly_stats.index,
                y=self.hourly_stats['mean'],
                name='Mean',
                line=dict(color='royalblue'),
            row=1, col=1
        ))
        
        fig.add_trace(
            go.Scatter(
                x=self.hourly_stats.index,
                y=self.hourly_stats['mean'] + 1.96*self.hourly_stats['std'],
                fill=None,
                mode='lines',
                line=dict(width=0),
            row=1, col=1
        ))
        
        fig.add_trace(
            go.Scatter(
                x=self.hourly_stats.index,
                y=self.hourly_stats['mean'] - 1.96*self.hourly_stats['std'],
                fill='tonexty',
                mode='lines',
                line=dict(width=0),
            row=1, col=1
        ))
        
        # P-values
        fig.add_trace(
            go.Bar(
                x=self.hourly_stats.index,
                y=-np.log10(self.hourly_stats['p_value']),
                name='-log10(p-value)',
                marker_color=np.where(self.hourly_stats['p_value'] < 0.05, 'red', 'grey')),
            row=2, col=1
        )
        
        fig.update_layout(
            height=600,
            title_text=f"Intraday {metric.capitalize()} Patterns",
            hovermode="x unified"
        )
        
        # Add session annotations
        for session, (start, end) in self.session_hours.items():
            fig.add_vrect(
                x0=start, x1=end,
                fillcolor="lightgray", opacity=0.25,
                line_width=0, row=1, col=1
            )
            fig.add_annotation(
                x=(start+end)/2, y=max(self.hourly_stats['mean']),
                text=session.capitalize(),
                showarrow=False,
                row=1, col=1
            )
        
        fig.add_hline(y=-np.log10(0.05), line_dash="dot", 
                     annotation_text="p=0.05", 
                     annotation_position="bottom right",
                     row=2, col=1)
        
        fig.show()
    
    def plot_session_heatmap(self):
        """Heatmap of returns by hour and day of week"""
        pivot = self.data.pivot_table(
            index='hour',
            columns='day_of_week',
            values='returns',
            aggfunc='mean'
        )
        
        days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
        
        fig = px.imshow(
            pivot,
            labels=dict(x="Day of Week", y="Hour", color="Returns"),
            x=days,
            y=pivot.index,
            color_continuous_scale='RdBu',
            aspect="auto"
        )
        
        # Add session rectangles
        for session, (start, end) in self.session_hours.items():
            fig.add_shape(
                type="rect",
                x0=-0.5, x1=6.5,
                y0=start, y1=end,
                line=dict(color="Black", width=2),
                fillcolor="LightGray",
                opacity=0.2
            )
        
        fig.update_layout(
            title="Hourly Returns by Day of Week",
            xaxis_nticks=7,
            yaxis_nticks=24
        )
        
        fig.show()
    
    def plot_seasonal_decomposition(self):
        """STL decomposition plot"""
        if not hasattr(self, 'seasonal'):
            self.decompose_intraday_seasonality()
            
        fig = make_subplots(rows=4, cols=1, shared_xaxes=True)
        
        # Original series
        fig.add_trace(
            go.Scatter(
                x=self.data.index,
                y=self.data['close'],
                name='Original'),
            row=1, col=1
        )
        
        # Trend component
        fig.add_trace(
            go.Scatter(
                x=self.data.index,
                y=self.trend,
                name='Trend'),
            row=2, col=1
        )
        
        # Seasonal component
        fig.add_trace(
            go.Scatter(
                x=self.data.index,
                y=self.seasonal,
                name='Seasonal'),
            row=3, col=1
        )
        
        # Residuals
        fig.add_trace(
            go.Scatter(
                x=self.data.index,
                y=self.resid,
                name='Residual'),
            row=4, col=1
        )
        
        fig.update_layout(
            height=800,
            title_text="STL Decomposition of Intraday Seasonality",
            hovermode="x unified"
        )
        
        fig.show()

# Example Usage
if __name__ == "__main__":
    # Load sample data (replace with your actual OHLC data)
    np.random.seed(42)
    date_rng = pd.date_range('2023-01-01', '2023-06-30', freq='5min')
    prices = np.cumprod(1 + np.random.normal(0, 0.0002, len(date_rng))) * 1.10
    
    # Add intraday seasonality pattern
    hour_effect = np.sin(np.pi * date_rng.hour / 24) * 0.0015
    prices = prices * (1 + hour_effect)
    
    data = pd.DataFrame({
        'open': prices,
        'high': prices + np.random.uniform(0, 0.0005, len(prices)),
        'low': prices - np.random.uniform(0, 0.0005, len(prices)),
        'close': prices
    }, index=date_rng)
    
    # Initialize analyzer
    analyzer = IntradayTimePatterns(data)
    
    # 1. Analyze and plot hourly returns
    hourly_stats = analyzer.analyze_hourly_patterns('returns')
    analyzer.plot_hourly_metrics('returns')
    
    # 2. Analyze session performance
    session_stats, anova_p = analyzer.analyze_session_patterns()
    print(f"\nSession Performance ANOVA p-value: {anova_p:.4f}")
    print(session_stats)
    
    # 3. Heatmap visualization
    analyzer.plot_session_heatmap()
    
    # 4. Seasonality decomposition
    analyzer.decompose_intraday_seasonality()
    analyzer.plot_seasonal_decomposition()
````
